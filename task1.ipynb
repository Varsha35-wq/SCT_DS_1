```python
import matplotlib.pyplot as plt

# Choose the year
year = '2020'

# Extract data
population_data = df[['Country Name', year]].dropna()

# Convert population values to numeric (in case they're not)
population_data[year] = pd.to_numeric(population_data[year], errors='coerce')

# Plot histogram
plt.figure(figsize=(12,6))
plt.hist(population_data[year], bins=30, color='orange', edgecolor='black')
plt.title(f'Distribution of Population Across Countries in {year}')
plt.xlabel('Population')
plt.ylabel('Number of Countries')
plt.grid(True)
plt.show()

```

```python
# Sort countries by population in descending order
top10 = population_data.sort_values(by=year, ascending=False).head(10)

# Plot bar chart
plt.figure(figsize=(10,6))
plt.bar(top10['Country Name'], top10[year], color='teal', edgecolor='black')
plt.title(f'Top 10 Most Populated Countries in {year}')
plt.xlabel('Country')
plt.ylabel('Population')
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```
